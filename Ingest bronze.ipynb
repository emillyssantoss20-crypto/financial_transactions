## 1. Configuração do ambiente

Define o caminho de onde estão os arquivos originais e seleciona a onde as tabelas da Bronze serão criadas.

```python
from pyspark.sql import functions as F

caminho = "/Volumes/financial_transactions/bronze/arquivos"
spark.sql("USE financial_transactions.bronze")
```

## 2. Clientes
Lê o arquivo de clientes, adiciona as colunas de controle e salva como tabela `users`.

```python
df_users = spark.read.csv(f"{caminho}/users_data.csv", header=True)

df_users = (df_users
    .withColumn("_arquivo_origem", F.col("_metadata.file_path"))
    .withColumn("_data_ingestao", F.current_timestamp()))

df_users.write.mode("overwrite").saveAsTable("users")
display(df_users)
```

## 3. Cartões
Lê o arquivo de cartões, adiciona as colunas de controle e salva como tabela `cards`.

```python
df_cards = spark.read.csv(f"{caminho}/cards_data.csv", header=True)

df_cards = (df_cards
    .withColumn("_arquivo_origem", F.col("_metadata.file_path"))
    .withColumn("_data_ingestao", F.current_timestamp()))

df_cards.write.mode("overwrite").saveAsTable("cards")
display(df_cards)
```

## 4. Transações
Lê o arquivo de transações, adiciona as colunas de controle e salva como tabela `transactions`.

```python
df_transactions = spark.read.csv(f"{caminho}/transactions_data.csv", header=True)

df_transactions = (df_transactions
    .withColumn("_arquivo_origem", F.col("_metadata.file_path"))
    .withColumn("_data_ingestao", F.current_timestamp()))

df_transactions.write.mode("overwrite").saveAsTable("transactions")
display(df_transactions)
```

## 5. Códigos MCC
Transforma o JSON de categorias de comércio, que vem como um único bloco de pares código e descrição, em uma tabela com uma linha por código.

```python
df_mcc = (spark.read.text(f"{caminho}/mcc_codes.json", wholetext=True)
    .selectExpr("explode(from_json(value, 'map<string,string>')) as (mcc, descricao)", "_metadata.file_path as _arquivo_origem")
    .withColumn("_data_ingestao", F.current_timestamp()))

df_mcc.write.mode("overwrite").saveAsTable("mcc_codes")
display(df_mcc)
```

## 6. Rótulos de fraude
Transforma o JSON de rótulos em uma tabela com uma linha por transação, indicando se ela é fraude ("Yes") ou não ("No").

```python
df_fraud = (spark.read.json(f"{caminho}/train_fraud_labels.json", multiLine=True, schema="target map<string,string>")
    .selectExpr("explode(target) as (transaction_id, is_fraud)", "_metadata.file_path as _arquivo_origem")
    .withColumn("_data_ingestao", F.current_timestamp()))

df_fraud.write.mode("overwrite").saveAsTable("fraud_labels")
display(df_fraud)
```

## 7. Conferência
Conta as linhas de cada tabela para confirmar que a carga foi concluída.

```python
%sql
SELECT 'users' AS tabela, COUNT(*) AS linhas FROM users
UNION ALL SELECT 'cards', COUNT(*) FROM cards
UNION ALL SELECT 'transactions', COUNT(*) FROM transactions
UNION ALL SELECT 'mcc_codes', COUNT(*) FROM mcc_codes
UNION ALL SELECT 'fraud_labels', COUNT(*) FROM fraud_labels
```
